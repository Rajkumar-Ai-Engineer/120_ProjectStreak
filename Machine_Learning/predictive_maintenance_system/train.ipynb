```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.metrics import classification_report,accuracy_score
from sklearn.neural_network import MLPClassifier
from imblearn.over_sampling import SMOTE
from sklearn.utils import estimator_html_repr
```

```python
data = pd.read_csv("ai4i2020.csv")
```

```python
data
```

# **Frequency Encoding**

```python
frequency = data['Product ID'].value_counts().to_dict()

data['Encoded_PID'] = data['Product ID'].map(frequency)
```

```python
data
```

# **TargetEncoding**

```python
'''target = data.groupby(['Product ID'])['Machine failure'].mean()
data['Encoded_PID'] = data['Product ID'].map(target)'''
```

# **Label Encoder**

```python
encoder = LabelEncoder()
data['Type_Encoded'] = encoder.fit_transform(data['Type'])
```

```python
data
```

```python
X = data.drop(['Product ID','Type','Machine failure',"UDI"],axis=1).values
```

```python
y = data['Machine failure'].values
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=None)
```

```python
resampler = SMOTE()
X_train,y_train =  resampler.fit_resample(X_train,y_train)

```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = MLPClassifier(hidden_layer_sizes=64,activation='relu',solver='adam',learning_rate='constant',verbose=True,learning_rate_init=0.001)

model.fit(X_train,y_train)
```

```python
pred_y = model.predict(X_test)
```

```python
acc = accuracy_score(y_test,pred_y)
```

```python
acc*100
```

```python
c_r = classification_report(y_test,pred_y)
```

```python
import pandas as pd

def create_machine_data():
    """
    Prompts the user to input values for machine data columns and returns a DataFrame.
    """
    print("Please provide the following details:")
    
    try:
        # Float64 inputs
        air_temperature = float(input("Air temperature [K] (e.g., 298.15): "))
        process_temperature = float(input("Process temperature [K] (e.g., 305.00): "))
        torque = float(input("Torque [Nm] (e.g., 40.5): "))

        # Int64 inputs
        rotational_speed = int(input("Rotational speed [rpm] (e.g., 1500): "))
        tool_wear = int(input("Tool wear [min] (e.g., 120): "))
        
        twf = int(input("TWF (Tool Wear Failure, 1 for Yes, 0 for No): "))
        hdf = int(input("HDF (Heat Dissipation Failure, 1 for Yes, 0 for No): "))
        pwf = int(input("PWF (Power Failure, 1 for Yes, 0 for No): "))
        osf = int(input("OSF (Overstrain Failure, 1 for Yes, 0 for No): "))
        rnf = int(input("RNF (Random Failures, 1 for Yes, 0 for No): "))
        encoded_pid = int(input("Encoded_PID (e.g., 101): "))

        # Int32 input
        type_encoded = int(input("Type_Encoded (e.g., 2): "))

        # Creating the DataFrame
        data = {
            'Air temperature [K]': [air_temperature],
            'Process temperature [K]': [process_temperature],
            'Rotational speed [rpm]': [rotational_speed],
            'Torque [Nm]': [torque],
            'Tool wear [min]': [tool_wear],
            
            'TWF': [twf],
            'HDF': [hdf],
            'PWF': [pwf],
            'OSF': [osf],
            'RNF': [rnf],
            'Encoded_PID': [encoded_pid],
            'Type_Encoded': [type_encoded],
        }

        # Setting data types explicitly
        df = pd.DataFrame(data)
        df = df.astype({
            'Air temperature [K]': 'float64',
            'Process temperature [K]': 'float64',
            'Rotational speed [rpm]': 'int64',
            'Torque [Nm]': 'float64',
            'Tool wear [min]': 'int64',
            
            'TWF': 'int64',
            'HDF': 'int64',
            'PWF': 'int64',
            'OSF': 'int64',
            'RNF': 'int64',
            'Encoded_PID': 'int64',
            'Type_Encoded': 'int32',
        })
        
        print("\nDataFrame created successfully:\n")
        print(df)
        return df

    except ValueError:
        print("\nError: Please ensure all inputs are of the correct type (float or int).")
        return None

# Example usage
df = create_machine_data()
```
