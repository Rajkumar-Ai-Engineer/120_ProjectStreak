```python
import mediapipe as mp 
import cv2 as cv
import numpy as np


hand_function = mp.solutions.hands
draw_img = mp.solutions.drawing_utils
hand_module = hand_function.Hands()


cam = cv.VideoCapture(0)

while True:
    ret, frame = cam.read()
    
    rgb = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
    result = hand_module.process(rgb)
    
    if result.multi_hand_landmarks:
        for hand in result.multi_hand_landmarks:
            draw_img.draw_landmarks(frame,hand,hand_function.HAND_CONNECTIONS,draw_img.DrawingSpec(color=(0, 255, 0), thickness=2, circle_radius=4),  # Custom landmark style
                                       draw_img.DrawingSpec(color=(255, 0, 0), thickness=2))
    
    cv.imshow("frame",frame)
    if cv.waitKey(1) & 0xff == ord('q'):
        break
    
cam.release()
cv.destroyAllWindows()
    
    
```
