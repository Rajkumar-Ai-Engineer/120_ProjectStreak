## 2.Import Necessary Libs

```python
from transformers import XLMRobertaTokenizer, XLMRobertaForSequenceClassification
from transformers import TrainingArguments, Trainer
from dataset import load_dataset
```

## 3.initialize model_id,model,Tokenizer

```python
model_id = "xlm-roberta-base"
model = XLMRobertaForSequenceClassification.from_pretrained(model_id)
tokenizer = XLMRobertaTokenizer.from_pretrained(model_id)
```

## 4.Load the dataset

```python
dataset_id = "go_emotions"
dataset = load_dataset(dataset_id,split="train")
```

## 5.Tokenize data

```python
def tokenize(batch):
    return tokenizer(batch['text'],padding=True,truncation=True)
tokenized_dataset = dataset.map(tokenize, batched = True)
```

## 6.Training Arguments

```python
training_args = TrainingArguments(
            output_dir = "./xberta-emotion",
            evaluation_strategy = "epoch",
            per_device_train_batch_size = 16,
            per_device_eval_batch_size = 32,
            num_train_epochs = 3,
            save_strategy = "epoch",
            logging_dir = "./logs"
)
```

## 7.Trainer

```python
trainer = Trainer(
            model = model,
            args = training_args,
            train_dataset = tokenized_dataset,
            eval_dataset = tokenized_dataset.select(range(1000)))
```

## 8.Training

```python
trainer.train()
```

## 9.Save the model

```python
trainer.save("xberta-emotion-model")
```
